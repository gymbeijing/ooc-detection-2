```python
def normalize_coordinates(coordinates, new_min=0, new_max=1):
    coord_min = coordinates.min(axis=0)
    coord_max = coordinates.max(axis=0)
    normalized = (coordinates - coord_min) / (coord_max - coord_min) * (new_max - new_min) + new_min
    return normalized
```
